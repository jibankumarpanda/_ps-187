# Face Recognition Demo

This demo uses the six registered images in `ml/data/faces` and the pretrained ArcFace `w600k_r50.onnx` model. The unknown-face example is decoded directly from the Kaggle ZIP and is never copied into the gallery.

```python
from pathlib import Path
import io
import zipfile
import cv2

from src.face_recognition import FaceRecognizer

ML_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
GALLERY = ML_ROOT / 'data' / 'faces'
recognizer = FaceRecognizer(gallery_dir=GALLERY, threshold=0.45)
print('registered identities:', sorted(recognizer.gallery))
```

```python
registered_image = cv2.imread(str(GALLERY / 'Alia_Bhatt.jpg'))
registered_result = recognizer.analyze(registered_image)
print('registered result:', registered_result)
assert registered_result[0]['identity'] == 'Alia_Bhatt'
```

```python
zip_path = Path.home() / 'Downloads' / 'archive.zip'
unknown_entry = 'Dataset/Faces/Billie Eilish/Billie Eilish_0.jpg'
with zipfile.ZipFile(zip_path) as archive:
    unknown_image = cv2.imdecode(__import__('numpy').frombuffer(archive.read(unknown_entry), __import__('numpy').uint8), cv2.IMREAD_COLOR)
unknown_result = recognizer.analyze(unknown_image)
print('unknown result:', unknown_result)
assert unknown_result[0]['identity'] == 'UNKNOWN'
```
